Step 1: Import Libraries

```python
import pandas as pd
import numpy as np
import random
```

Step 2: Define Domain Values

```python
np.random.seed(42)

departments = {
    "Cardiology": ["Heart Failure", "Hypertension"],
    "General Medicine": ["Diabetes", "Infection"],
    "Orthopedics": ["Fracture"],
    "Neurology": ["Stroke", "Migraine"],
    "Geriatrics": ["Stroke", "Dementia"]
}

genders = ["Male", "Female"]
admission_types = ["Emergency", "Planned"]
discharge_options = ["Home", "Rehab"]
```

Step 3: Readmission Logic (REALISTIC ⭐)

```python
def assign_readmission(age, stay, meds):
    risk = 0
    if age > 60:
        risk += 0.3
    if stay > 5:
        risk += 0.3
    if meds > 7:
        risk += 0.2
    return "Yes" if random.random() < risk else "No"
```

Step 4: Generate Data

```python
rows = []

for i in range(1000):
    dept = random.choice(list(departments.keys()))
    diagnosis = random.choice(departments[dept])
    
    age = np.random.randint(20, 90)
    gender = random.choice(genders)
    length_of_stay = np.random.randint(1, 15)
    num_medications = np.random.randint(1, 12)
    num_procedures = np.random.randint(0, 5)
    admission = random.choice(admission_types)
    discharge = random.choice(discharge_options)
    
    readmitted = assign_readmission(age, length_of_stay, num_medications)

    rows.append([
        f"P{i+1:04}",
        age,
        gender,
        dept,
        diagnosis,
        length_of_stay,
        num_medications,
        num_procedures,
        admission,
        discharge,
        readmitted
    ])
```

Step 5: Create DataFrame & Save CSV

```python
columns = [
    "patient_id", "age", "gender", "department", "diagnosis",
    "length_of_stay", "num_medications", "num_procedures",
    "admission_type", "discharge_disposition", "readmitted"
]

df = pd.DataFrame(rows, columns=columns)

df.to_csv("hospital_readmission_data_1000.csv", index=False)

df.head()
```
